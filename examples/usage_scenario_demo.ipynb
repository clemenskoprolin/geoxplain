# GeoXplain: explaining humidity over Zurich

```python
import geoxplain_aurora_adapter as ax
from geoxplain import GeoXplainWidget

REMOTE = "http://localhost:8765" # run remotely
# REMOTE = None
```

```python
target = ax.Target.box(          # what do we want to explain?
    var="q",
    level=850,
    lat=47.38, lon=8.54,         # Zurich
    timestamp="2024-07-15T12:00:00Z",
    size=(1.0, 1.6),
)
```

```python
ig_result = ax.run_ig(           # how do we explain it?
    target=target,
    input=["t", "q"],
    remote=REMOTE,
)
```

## Viewer

```python
widget = GeoXplainWidget(
    height=680,
    # title="Zurich humidity forecast",    # optional
    # subtitle="Aurora · 850 hPa",         # optional
)

widget.add_attribution(ig_result)
widget
```

```python
rise_result = ax.run_rise(
    target=target,
    input=["t"],
    remote=REMOTE,
)

# widget.add_attribution(rise_result)
```

## Add meteorological context

```python
humidity_overlay = ax.pull_overlay(
    "q",
    level=850,        # other metadata like inferred
    remote=REMOTE,
)

widget.add_overlay(humidity_overlay)
```

```python
temperature_overlay = ax.pull_overlay(
    "t",
    level=850,
    remote=REMOTE,
)

widget.add_overlay(temperature_overlay)
```

## Autoregressive rollout

```python
rollout_result = ax.run_rollout(
    target=target,
    input=["t"],
    method="ig",
    timeframes=4,
    remote=REMOTE,
)

# widget.add_attribution(rollout_result)
```

```python
U_WIND_COLORMAP = [
    (0.0, "#234f8c"),
    (0.5, "#f7f7f7"),
    (1.0, "#f28e2b"),
]

u_wind_overlay = ax.pull_overlay(
    "u",
    level=850,
    remote=REMOTE,
)
widget.add_overlay(u_wind_overlay, colormap=U_WIND_COLORMAP, opacity=0.55)
```

```python
V_WIND_COLORMAP = [
    (0.0, "#762a83"),
    (0.5, "#f7f7f7"),
    (1.0, "#1b9e77"),
]

v_wind_overlay = ax.pull_overlay(
    "v",
    level=850,
    remote=REMOTE,
)
widget.add_overlay(v_wind_overlay, colormap=V_WIND_COLORMAP, opacity=0.55)
```
